```python

%%writefile expense_logger.py
"""A simple terminal-based expense logger."""

def add(ledger, amount, category, note=""):
    ledger.append({
        "amount": amount,
        "category": category,
        "note": note
    })


def balance(ledger):
    total = 0
    for entry in ledger:
        total += entry["amount"]
    return total


def show(ledger):
    for number, entry in enumerate(ledger, start=1):
        print(
            f"{number:>3}  {entry['amount']:>10.2f}  "
            f"{entry['category']:<12} {entry['note']}"
        )
    print(f"     {'balance':>10}  {balance(ledger):>10.2f}")


def main():
    ledger = []
    print("commands: add <amount> <category> [note] | list | balance | quit")
    print("income is positive, an expense is negative")

    while True:
        try:
            line = input().strip()
        except EOFError:
            break

        if line == "":
            continue

        words = line.split()
        command = words[0]

        if command == "add" and len(words) >= 3:
            add(ledger, float(words[1]), words[2], " ".join(words[3:]))
        elif command == "list":
            show(ledger)
        elif command == "balance":
            print(balance(ledger))
        elif command == "quit":
            break
        else:
            print("sorry, I don't understand that")


if __name__ == "__main__":
    main()
```

```python
%%writefile mybudget/account.py
"""The account (lessons 2-3)."""

from __future__ import annotations

import datetime as dt
from decimal import Decimal

from .transaction import Transaction


class Account:
    def __init__(self, name: str = "Main"):
        self.name = name
        self._transactions: list[Transaction] = []

    @property
    def balance(self) -> Decimal:
        return sum((t.amount for t in self._transactions), Decimal(0))

    @property
    def transactions(self) -> tuple[Transaction, ...]:
        return tuple(self._transactions)

    def add(self, transaction: Transaction, until: dt.date | None = None) -> list[Transaction]:
        """Add a transaction (or, for a recurring one, every occurrence up to ``until``)."""
        return self.add_many(transaction.occurrences(until))

    def add_many(self, transactions) -> list[Transaction]:
        batch = sorted(transactions, key=lambda t: t.date)
        self._transactions.extend(batch)
        return batch
```

```python
%%writefile mybudget/reports.py
"""Reports (lesson 3): abstraction."""

from __future__ import annotations

from abc import ABC, abstractmethod
from decimal import Decimal

from .account import Account


class Report(ABC):
    title = "Report"

    @abstractmethod
    def generate(self, account: Account) -> list[tuple[str, Decimal]]:
        """Return (label, amount) rows. Each subclass decides what the rows mean."""

    def render(self, account: Account) -> str:
        rows = self.generate(account)
        lines = [self.title, "-" * len(self.title)]
        lines += [f"{label:<20}{amount:>14,.2f}" for label, amount in rows]
        if not rows:
            lines.append("(no transactions)")
        return "\n".join(lines)
```

```python
from pathlib import Path
from mybudget import load

bad_path = Path("corrupted_test.json")
bad_path.write_text("{this is not valid JSON", encoding="utf-8")

result = load(bad_path)

print("Transactions:", len(result.account.transactions))
print("Problems:", result.problems)
print("Original file exists:", bad_path.exists())
print("Backup of corrupted file exists:", Path("corrupted_test.json.corrupt").exists())
```

```python
import json
from pathlib import Path
from mybudget import load

partial_path = Path("partial_test.json")

test_data = {
    "version": 1,
    "name": "Recovery Test",
    "allow_overdraft": True,
    "budgets": {},
    "transactions": [
        {
            "date": "2026-10-01",
            "amount": "3000",
            "category": "salary",
            "description": "October salary"
        },
        {
            "date": "not-a-date",
            "amount": "-500",
            "category": "rent",
            "description": "Invalid date"
        },
        {
            "date": "2026-10-03",
            "amount": "-200",
            "category": "food",
            "description": "Groceries"
        }
    ]
}

partial_path.write_text(json.dumps(test_data, indent=2), encoding="utf-8")

result = load(partial_path)

print("Account:", result.account.name)
print("Transactions loaded:", len(result.account.transactions))
print("Balance:", result.account.balance)
print("Problems:", result.problems)
```

```python
from pathlib import Path
from mybudget import Account, Transaction, save, save_with_backup, load

backup_test = Path("backup_test.json")
backup_file = Path("backup_test.json.bak")

# Create and save the original account.
old_account = Account("Original")
old_account.add(Transaction.parse("2026-10-01", "1000", "salary"))
save(old_account, backup_test)

# Save a different account using the backup feature.
new_account = Account("Updated")
new_account.add(Transaction.parse("2026-10-02", "2000", "salary"))
save_with_backup(new_account, backup_test)

print("Backup exists:", backup_file.exists())
print("Current account:", load(backup_test).account.name)
print("Backup account:", load(backup_file).account.name)
```

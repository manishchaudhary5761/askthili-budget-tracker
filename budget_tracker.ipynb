```python

# Budget Tracker - Version 2

income = float(input("Enter your monthly income (₹): "))

expenses = []

number_of_expenses = int(input("How many expenses do you want to enter? "))

for i in range(number_of_expenses):
    expense = float(input(f"Enter expense {i + 1} (₹): "))
    expenses.append(expense)

total_expenses = sum(expenses)
balance = income - total_expenses

print("\n--- MONTHLY BUDGET REPORT ---")
print(f"Monthly Income: ₹{income:.2f}")
print(f"Total Expenses: ₹{total_expenses:.2f}")
print(f"Remaining Balance: ₹{balance:.2f}")


```

```python

# Budget Tracker - Version 3

income = float(input("Enter your monthly income (₹): "))

expenses = []

number_of_expenses = int(input("How many expenses do you want to enter? "))

for i in range(number_of_expenses):
    print(f"\nExpense {i + 1}")

    name = input("What did you spend money on? ")
    amount = float(input("How much did you spend (₹)? "))

    expense = {
        "name": name,
        "amount": amount
    }

    expenses.append(expense)

total_expenses = sum(expense["amount"] for expense in expenses)
balance = income - total_expenses

print("\n===== MONTHLY BUDGET REPORT =====")
print(f"Monthly Income: ₹{income:.2f}")
print("\nYour Expenses:")

for expense in expenses:
    print(f"{expense['name']}: ₹{expense['amount']:.2f}")

print("---------------------------------")
print(f"Total Expenses: ₹{total_expenses:.2f}")
print(f"Remaining Balance: ₹{balance:.2f}")
```

```python

%%writefile expense_logger.py
"""A simple terminal-based expense logger."""

def add(ledger, amount, category, note=""):
    ledger.append({
        "amount": amount,
        "category": category,
        "note": note
    })


def balance(ledger):
    total = 0
    for entry in ledger:
        total += entry["amount"]
    return total


def show(ledger):
    for number, entry in enumerate(ledger, start=1):
        print(
            f"{number:>3}  {entry['amount']:>10.2f}  "
            f"{entry['category']:<12} {entry['note']}"
        )
    print(f"     {'balance':>10}  {balance(ledger):>10.2f}")


def main():
    ledger = []
    print("commands: add <amount> <category> [note] | list | balance | quit")
    print("income is positive, an expense is negative")

    while True:
        try:
            line = input().strip()
        except EOFError:
            break

        if line == "":
            continue

        words = line.split()
        command = words[0]

        if command == "add" and len(words) >= 3:
            add(ledger, float(words[1]), words[2], " ".join(words[3:]))
        elif command == "list":
            show(ledger)
        elif command == "balance":
            print(balance(ledger))
        elif command == "quit":
            break
        else:
            print("sorry, I don't understand that")


if __name__ == "__main__":
    main()
```

```python

import subprocess
import sys

commands = """add 1000 salary
add -250.50 rent march
add -42.10 groceries
list
balance
quit
"""

result = subprocess.run(
    [sys.executable, "expense_logger.py"],
    input=commands,
    capture_output=True,
    text=True
)

print(result.stdout)

if result.returncode != 0:
    print("Error:", result.stderr)
```

```python

def by_category(ledger):
    totals = {}

    for entry in ledger:
        category = entry["category"]

        if category not in totals:
            totals[category] = 0

        totals[category] += entry["amount"]

    return totals


# Test the function with sample transactions
ledger = [
    {"amount": 1000.00, "category": "salary", "note": ""},
    {"amount": -250.50, "category": "rent", "note": "march"},
    {"amount": -42.10, "category": "groceries", "note": ""},
    {"amount": -9.99, "category": "groceries", "note": "milk"}
]

print(by_category(ledger))
```

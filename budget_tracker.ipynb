```python

%%writefile expense_logger.py
"""A simple terminal-based expense logger."""

def add(ledger, amount, category, note=""):
    ledger.append({
        "amount": amount,
        "category": category,
        "note": note
    })


def balance(ledger):
    total = 0
    for entry in ledger:
        total += entry["amount"]
    return total


def show(ledger):
    for number, entry in enumerate(ledger, start=1):
        print(
            f"{number:>3}  {entry['amount']:>10.2f}  "
            f"{entry['category']:<12} {entry['note']}"
        )
    print(f"     {'balance':>10}  {balance(ledger):>10.2f}")


def main():
    ledger = []
    print("commands: add <amount> <category> [note] | list | balance | quit")
    print("income is positive, an expense is negative")

    while True:
        try:
            line = input().strip()
        except EOFError:
            break

        if line == "":
            continue

        words = line.split()
        command = words[0]

        if command == "add" and len(words) >= 3:
            add(ledger, float(words[1]), words[2], " ".join(words[3:]))
        elif command == "list":
            show(ledger)
        elif command == "balance":
            print(balance(ledger))
        elif command == "quit":
            break
        else:
            print("sorry, I don't understand that")


if __name__ == "__main__":
    main()
```

```python
%%writefile mybudget/account.py
"""The account (lessons 2-3)."""

from __future__ import annotations

import datetime as dt
from decimal import Decimal

from .transaction import Transaction


class Account:
    def __init__(self, name: str = "Main"):
        self.name = name
        self._transactions: list[Transaction] = []

    @property
    def balance(self) -> Decimal:
        return sum((t.amount for t in self._transactions), Decimal(0))

    @property
    def transactions(self) -> tuple[Transaction, ...]:
        return tuple(self._transactions)

    def add(self, transaction: Transaction, until: dt.date | None = None) -> list[Transaction]:
        """Add a transaction (or, for a recurring one, every occurrence up to ``until``)."""
        return self.add_many(transaction.occurrences(until))

    def add_many(self, transactions) -> list[Transaction]:
        batch = sorted(transactions, key=lambda t: t.date)
        self._transactions.extend(batch)
        return batch
```

```python
%%writefile mybudget/reports.py
"""Reports (lesson 3): abstraction."""

from __future__ import annotations

from abc import ABC, abstractmethod
from decimal import Decimal

from .account import Account


class Report(ABC):
    title = "Report"

    @abstractmethod
    def generate(self, account: Account) -> list[tuple[str, Decimal]]:
        """Return (label, amount) rows. Each subclass decides what the rows mean."""

    def render(self, account: Account) -> str:
        rows = self.generate(account)
        lines = [self.title, "-" * len(self.title)]
        lines += [f"{label:<20}{amount:>14,.2f}" for label, amount in rows]
        if not rows:
            lines.append("(no transactions)")
        return "\n".join(lines)
```
